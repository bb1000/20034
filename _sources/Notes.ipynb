How to we translate

* k -> $10^3$
* M -> $10^6$
* B -> $10^9$

in a systematic way?

Consider 3.g. `10.6M` meaning $10.6*10^6$

A function that that takes the string and splits off the final character


````python
def convert(x: str) -> float:
    factors = {'k': 10**3, 'M': 10**6, 'B': 10**9}
    try:
        if x[-1] in factors:
            return float(x[:-1]) * factors[x[-1]]
        else:
            return float(x)
    except TypeError:
        # do nothing
        return x
        
````

````python
convert(2.0), convert('3k'), convert('4M'), convert('1B')
````

The buildin function `map` applies a function on a sequence of values

````python
for x in map(convert, [2.0, '3k', '4M', '1B']):
    print(x)
````

In a similar vein pandas has a map method that is applied to its members

````python
# Example: complex number

class MyComplex:
    """
    Implements complex numbers in Python
    """
    def __init__(self, re=0, im=0):
        self.re = float(re)
        self.im = float(im)

    def __repr__(self):
        return f"MyComplex({self.re}, {self.im})"

    def __str__(self):
        return f"{self.re} + {self.im}i"

    # def __add__(left, right)
    def __add__(self, other):
        new_z = MyComplex(self.re + other.re, self.im + other.im)
        return new_z
        
    
````

````python
# %load git_demo/savings.py
# Calculate retirement savings
import sys
import matplotlib.pyplot as plt

def savings_calculator(amount, interest, years=1, start_deposit=None, stop_deposit=None, plot=False):
    final_amount = 0
    savings_by_year = []
    if start_deposit is None:
        start_deposit = 0

    if stop_deposit is None:
        stop_deposit = years
        
    for year in range(years):
        if start_deposit <= year < stop_deposit:
            final_amount = (amount + final_amount)*(1 + interest)
        else:
            final_amount = final_amount*(1 + interest)

        final_amount =  round(final_amount, 2)
        #print(year, final_amount)
        savings_by_year.append(final_amount)

    if plot:
        plt.plot(savings_by_year)

        
    return final_amount
````

````python
t1 = time.time()
slow_math(3)
t2 = time.time()
print("Time used in slow_math", round(t2-t1, 1))
````

````python
# with a decorator - function that takes a function as input, and returns a function

def timeme(function):
    def timed_function(*args, **kwargs):
        t1 = time.time()
        return_value = function(*args, **kwargs)
        t2 = time.time()
        print("Time used", round(t2-t1, 1))
        return return_value
    return timed_function
````

````python
timeme(slow_math)
````

````python
slow_math_timed = timeme(slow_math)
````

````python
slow_math_timed(4)
````

````python
# decorator notation

@timeme  # equivalent to slow_math=timeme(slow_math), redefinition
def slow_math(x):
    time.sleep(1)
    return math.sqrt(x)
````

````python
slow_math(2)
````

````python
def debug(function):
    def wrapped(*args, **kwargs):
        print(function.__name__, 'called with', args, kwargs)
        return_value = function(*args, **kwargs)
        print(function.__name__, 'returns', return_value)
        return return_value
    return wrapped
````

````python
@debug  
def slow_math(x):
    time.sleep(1)
    return math.sqrt(x)
````

````python
slow_math(2)
````

````python
# combination of decorators
@timeme
@debug  
def slow_math(x):
    time.sleep(1)
    return math.sqrt(x)
````

````python
slow_math(2)
````

````python
# combination of decorators
@debug  
@timeme
def slow_math(x):
    time.sleep(1)
    return math.sqrt(x)
````

````python
slow_math(2)
````

````python
from functools import wraps

def timeme(function):
    @wraps(function)
    def timed_function(*args, **kwargs):
        t1 = time.time()
        return_value = function(*args, **kwargs)
        t2 = time.time()
        print("Time used in", function.__name__, round(t2-t1, 1))
        return return_value
    return timed_function

def debug(function):
    @wraps(function)
    def wrapped(*args, **kwargs):
        print(function.__name__, 'called with', args, kwargs)
        return_value = function(*args, **kwargs)
        print(function.__name__, 'returns', return_value)
        return return_value
    return wrapped
````

````python
# combinatin of decorators with wraps

@timeme
@debug
def slow_math(x):
    time.sleep(1)
    return math.sqrt(x)

slow_math(2)
````

````python
class TimeMe:
    def __enter__(self):
        self.t1 = time.time()

    def __exit__(self, *args):
        self.t2 = time.time()
        print("Time used in block", round(self.t2 - self.t1, 1))
        
````

````python
with TimeMe():
    time.sleep(2)
````

````python
def myrange(n):
    i = 0
    while i < n:
        yield i
        i += 1
````

````python
#Example: list the first n Fibonacci numbers:  0 1 1 2 3 5 8...

def fib(n):
    counter = 0
    a = 0
    b = 1
    while counter < n:
        yield a
        # b -> a
        # a + b -> b
        a, b = b, a + b
        counter += 1
    
    
````

````python
for fibnum in fib(5):
    print(fibnum)
````

````python
list(fib(20))
````

````python
def grep(search_string, filename):
    with open(filename) as f:
        for line in f:
            if search_string in line:
                yield line.strip('\n')
````

````python
for matching_line in grep('hello', 'data.txt'):
    print(matching_line)
````

````python
def tail(filename):
    with open(filename) as f:
        while True:
            line = f.readline()
            if not line:
                time.sleep(1)
                continue
            yield line
````

````python
for line in tail('data.txt'):
    print(line)
````

````python
# a more general way
f = open('data.txt')

def grep(search_string, f):
    for line in f:
       if search_string in line:
             yield line.strip('\n')
````

````python
for matching_line in grep('hello', f):
    print(matching_line)
````

````python
def tail(f):
    while True:
        line = f.readline()
        if not line:
            time.sleep(1)
            continue
        yield line
````

````python
f = open('data.txt')
for line in tail(f):
    print(line)
````

````python
for line in grep('python', tail(open('data.txt'))):
    print(line)
````
